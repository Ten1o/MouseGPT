```python
for period in periods:
    period_avg_velo = 0
    for group in period:
        with open(f'{root_folder}/{group}/velocity.json') as f:
            data = json.load(f)

        frame_num = len(data)
        group_avg_velo = 0
        for v in data.values():
            group_avg_velo += v['hindpawR'] / frame_num
    period_avg_velo += group_avg_velo/len(period)

    print(period_avg_velo)
```

```python

fig, axes = plt.subplots(4, 4, figsize=(15, 10))

fid = 0
for gid in range(len(periods[0])):
    group_velo = np.zeros(3)
    for pid in range(len(periods)):
        group = periods[pid][gid]
        with open(f'{root_folder}/{group}/velocity.json') as f:
            data = json.load(f)

        frame_num = len(data)
        group_avg_velo = 0
        for v in data.values():
            group_avg_velo += v['hindpawR'] / frame_num
        
        group_velo[pid] = group_avg_velo

    axes.flat[fid].bar(np.arange(3), group_velo)
    axes.flat[fid].set_title(group)
    fid += 1


plt.tight_layout()
plt.show()
```
