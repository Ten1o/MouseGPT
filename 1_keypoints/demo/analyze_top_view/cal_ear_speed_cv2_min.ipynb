```python
import json
import numpy as np
from tqdm import tqdm
import cv2

from scipy.signal import savgol_filter

import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.gridspec import GridSpec
```

```python
import os
print('Loading keypoints json ……')

json_path = os.environ['MOUSEGPT_KEYPOINTS_JSON']
with open(json_path, 'r') as f:
    data = json.load(f)
    
frame_num = len(data)
print(f'total {frame_num} frames')
```

```python
ears_pos = []

for frame in tqdm(data, desc='loading kpts'):
    ears = np.array(frame['instances'][0]['keypoints'][0:2])
    ears_pos.append(ears)

ears_pos = np.array(ears_pos)
print(ears_pos.shape)
```

```python
ear_speed = []

for index in tqdm(range(1, frame_num), desc='cal speed'):
    speed = np.linalg.norm(ears_pos[index]-ears_pos[index-1], axis=1)
    ear_speed.append(speed)

ear_speed = np.array(ear_speed)

# min speed of ear
ear_speed = np.concatenate([ear_speed, np.min(ear_speed, axis=1).reshape(-1,1)], axis=1)

print(ear_speed.shape)
print(ear_speed[:5])
```

```python
def filter(speed):
    # Apply the Savitzky-Golay filter to the 3D keypoint data
    window_length = 40
    poly_order = 7

    # Ensure window length is odd
    if window_length % 2 == 0:
        window_length += 1

    filtered_data_3d = np.empty_like(speed)

    # Apply the filter for each dimension separately
    for ear in range(speed.shape[1]):
            filtered_data_3d[:, ear] = savgol_filter(speed[:, ear], window_length, poly_order)
    return filtered_data_3d

filtered_ear_speed = filter(ear_speed)

print(filtered_ear_speed[:5])
```

```python

fig, axs = plt.subplots(3,1)

lines = []
for ax in axs:
    ax.set_xlabel('Frame')
    ax.set_ylabel('Speed')

    ax.set_ylim(0, min(40, filtered_ear_speed.max()))

    line, = ax.plot([], [], 'b-')
    lines.append(line)
```

```python
import os

cap = cv2.VideoCapture(os.environ['MOUSEGPT_VIDEO'])


fps = cap.get(cv2.CAP_PROP_FPS)


output_width = int(1680)
output_height = int(800)
output_fps = fps//2
output_filename = os.environ['MOUSEGPT_OUTPUT_VIDEO']
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_filename, fourcc, output_fps, (output_width, output_height))
```

```python
index = 0

while cap.isOpened():
    index += 1
    print(index, end='\r')

    if index % 2 == 0:
        ret = cap.grab()
        continue
    else:
        ret, frame = cap.read()
    if not ret:
        break
    

    end_index = min(frame_num-1, index + 2)
    start_index = end_index-600
    speed_start_index = max(0, start_index)
    
    # Ear 1
    current_speed_data = np.zeros(600)
    current_speed_data[max(0, 600-(end_index-speed_start_index)):600] = filtered_ear_speed[speed_start_index:end_index, 0]

    axs[0].set_xlim(max(start_index, 0), max(end_index, 600))
    

    lines[0].set_data(range(start_index, end_index), current_speed_data)
    
    # Ear 2
    current_speed_data = np.zeros(600)
    current_speed_data[max(0, 600-(end_index-speed_start_index)):600] = filtered_ear_speed[speed_start_index:end_index, 1]

    axs[1].set_xlim(max(start_index, 0), max(end_index, 600))
    

    lines[1].set_data(range(start_index, end_index), current_speed_data)
    
    # Ear 2
    current_speed_data = np.zeros(600)
    current_speed_data[max(0, 600-(end_index-speed_start_index)):600] = filtered_ear_speed[speed_start_index:end_index, 2]

    axs[2].set_xlim(max(start_index, 0), max(end_index, 600))
    

    lines[2].set_data(range(start_index, end_index), current_speed_data)

    fig.canvas.draw()
    plt.pause(0.001)
    

    plt_img_min = np.frombuffer(fig.canvas.tostring_rgb(), dtype=np.uint8)
    plt_img_min = plt_img_min.reshape(fig.canvas.get_width_height()[::-1] + (3,))
    plt_img_min = cv2.cvtColor(plt_img_min, cv2.COLOR_RGB2BGR)
    plt_img_min = cv2.resize(plt_img_min, (400,800))


    combined_frame = np.hstack([plt_img_min, frame])

    out.write(combined_frame)
 
    # if index > 5000:
    #     break


cap.release()
out.release()
```

```python
from scipy.signal import find_peaks

peaks, infos = find_peaks(filtered_ear_speed[:, 2], height=5)

drop = []

for i in range(1, len(peaks)):
    if peaks[i] - peaks[i-1] <= 240:
        drop.append(i)
        drop.append(i-1)
peaks = [peaks[i] for i in range(len(peaks)) if i not in drop]

print(len(peaks))
```
